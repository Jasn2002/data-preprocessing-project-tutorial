```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#importing the data
df = pd.read_csv('AB_NYC_2019.csv')
#dropping duplicates
df = df.drop_duplicates(subset=['id'], keep='first', inplace=False)
#dropping irrelevant information
df = df.drop(columns=['id', 'name', 'host_id', 'host_name'], inplace=False)
```

```python
print(df.shape)
```

```python
print(df.info())
```

```python
fig, axis = plt.subplots(2, 3, figsize = (20, 7))

# Create a multiple histogram
sns.histplot(ax = axis[0, 0], data = df, x = "neighbourhood").set(ylabel = None)
sns.histplot(ax = axis[0, 1], data = df, x = "neighbourhood_group").set(ylabel = None)
sns.histplot(ax = axis[0, 2], data = df, x = "room_type").set(ylabel = None)
sns.histplot(ax = axis[1, 0], data = df, x = "last_review")
sns.histplot(ax = axis[1, 1], data = df, x = "reviews_per_month").set(ylabel = None)
sns.histplot(ax = axis[1, 2], data = df, x = "availability_365").set(ylabel = None)

# Adjust the layout
plt.tight_layout()

# Show the plot
plt.show()
```

```python
fig, axis = plt.subplots(4, 2, figsize = (10, 7), gridspec_kw={'height_ratios': [6, 1, 6, 1]})

# Creating a multiple figure with histograms and box plots
sns.histplot(ax = axis[0, 0], data = df, x = "price").set(xlabel = None)
sns.boxplot(ax = axis[1, 0], data = df, x = "price")
sns.histplot(ax = axis[0, 1], data = df, x = "minimum_nights").set(xlabel = None, ylabel = None)
sns.boxplot(ax = axis[1, 1], data = df, x = "minimum_nights")
sns.histplot(ax = axis[2, 0], data = df, x = "number_of_reviews").set(xlabel = None)
sns.boxplot(ax = axis[3, 0], data = df, x = "number_of_reviews")
sns.histplot(ax = axis[2, 1], data = df, x = "calculated_host_listings_count").set(xlabel = None, ylabel = None)
sns.boxplot(ax = axis[3, 1], data = df, x = "calculated_host_listings_count")

# Adjust the layout
plt.tight_layout()

# Show the plot
plt.show()
```

```python
fig, axis = plt.subplots(2, 2, figsize = (10, 7))

# Create a multiple scatter diagram
sns.regplot(ax = axis[0, 0], data = df, x = "minimum_nights", y = "price")
sns.heatmap(df[["price", "minimum_nights"]].corr(), annot = True, fmt = ".2f", ax = axis[1, 0], cbar = False)
sns.regplot(ax = axis[0, 1], data = df, x = "number_of_reviews", y = "price").set(ylabel=None)
sns.heatmap(df[["price", "number_of_reviews"]].corr(), annot = True, fmt = ".2f", ax = axis[1, 1])

# Adjust the layout
plt.tight_layout()

# Show the plot
plt.show()
```

```python
fig, axis = plt.subplots(figsize = (15, 7))

sns.countplot(data = df, x = "room_type", hue = "neighbourhood_group")

plt.tight_layout()
plt.show()
```
